# We can also add some additional columns to indicate how long a store has been running promo2 and whether a new round of promo2 starts in the current month.

```python
def check_promo_month(row):
    month2str = {1:'Jan', 2:'Feb', 3:'Mar', 4:'Apr', 5:'May', 
                 6:'Jun', 7:'Jul', 8:'Aug', 9:'Sept', 10:'Oct', 
                 11:'Nov',12:'Dec'}
    try:
        months = (row['PromoInterval'] or '').split(',')
        if row['Promo2Open'] and month2str[row['Month']] in months:
            return 1
        else:
            return 0
    except Exception:
        return 0    
  
def promo_cols(df):
    # Months since Promo2 started
    promo2_year = df['Promo2SinceYear'].fillna(0)
    promo2_week = df['Promo2SinceWeek'].fillna(0)
    df['Promo2Open'] = (
        12 * (df['Year'] - promo2_year)
        + (df['WeekOfYear'] - promo2_week) / 4
    ).clip(lower=0) * df['Promo2']
    # Whether a new round of promotions was started in the current month
    df['IsPromo2Month'] = df.apply(check_promo_month, axis=1) * df['Promo2']
```

```python
def mean_squared_percentage_error(y_true, y_pred):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    nonzero = y_true != 0
    return np.mean(((y_true[nonzero] - y_pred[nonzero]) / y_true[nonzero]) ** 2)
```

```python
import re
import textwrap
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch


def plot_tree_left_to_right(model):
    tree_dot = model.get_booster().get_dump(dump_format="dot")[0]
    nodes, children, edges = {}, {}, []
    node_pattern = re.compile(r'^\s*(\d+) \[ label="(.*?)" \]')
    edge_pattern = re.compile(r'^\s*(\d+) -> (\d+) \[label="(.*?)"')

    for line in tree_dot.splitlines():
        node_match = node_pattern.match(line)
        edge_match = edge_pattern.match(line)
        if node_match:
            node_id, label = node_match.groups()
            nodes[node_id] = label
            children.setdefault(node_id, [])
        elif edge_match:
            parent, child, branch = edge_match.groups()
            children.setdefault(parent, []).append(child)
            children.setdefault(child, [])
            edges.append((parent, child, branch))

    positions, leaf_position = {}, [0]

    def place_node(node_id, depth=0):
        child_nodes = children[node_id]
        if child_nodes:
            child_positions = [place_node(child_id, depth + 1) for child_id in child_nodes]
            y_position = sum(child_positions) / len(child_positions)
        else:
            y_position = leaf_position[0] * 1.35
            leaf_position[0] += 1
        positions[node_id] = (depth * 3.2, y_position)
        return y_position

    root_id = next(node_id for node_id in nodes if node_id not in {child for _, child, _ in edges})
    place_node(root_id)
    max_x = max(x for x, _ in positions.values())
    max_y = max(y for _, y in positions.values())
    fig, ax = plt.subplots(figsize=(max(14, max_x + 4), max(9, max_y + 2)))

    for parent, child, branch in edges:
        parent_x, parent_y = positions[parent]
        child_x, child_y = positions[child]
        ax.annotate("", xy=(child_x, child_y), xytext=(parent_x, parent_y),
                    arrowprops={"arrowstyle": "-|>", "color": "#64748b", "lw": 1.2})
        ax.text((parent_x + child_x) / 2, (parent_y + child_y) / 2 + 0.12,
                branch, ha="center", va="bottom", fontsize=9, color="#475569")

    for node_id, (x_position, y_position) in positions.items():
        label = textwrap.fill(nodes[node_id], width=24)
        face_color = "#dcfce7" if not children[node_id] else "#dbeafe"
        box = FancyBboxPatch((x_position - 1.05, y_position - 0.38), 2.1, 0.76,
                             boxstyle="round,pad=0.04,rounding_size=0.06",
                             facecolor=face_color, edgecolor="#334155", linewidth=1.1)
        ax.add_patch(box)
        ax.text(x_position, y_position, label, ha="center", va="center", fontsize=8)

    ax.set_xlim(-1.4, max_x + 1.4)
    ax.set_ylim(-0.8, max_y + 0.8)
    ax.invert_yaxis()
    ax.axis("off")
    ax.set_title("XGBoost Decision Tree", fontsize=16, pad=18)
    plt.tight_layout()
    plt.show()
```

```python
from sklearn.model_selection import KFold
```

```python
# Helper Funtion
```

```python
def train_and_evaluate(X_train, train_targets, X_val, val_targets, **params):
    model = XGBRegressor(random_state=42, n_jobs=-1, **params)
    model.fit(X_train, train_targets)
    train_rmse = np.sqrt(mean_squared_error(model.predict(X_train), train_targets))
    val_rmse = np.sqrt(mean_squared_error(model.predict(X_val), val_targets))
    return model, train_rmse, val_rmse
```

```python
KFold = KFold(n_splits=5, shuffle=True)
```

```python
models = []
```

```python
for train_idxs, val_idxs in KFold.split(x_train):
    X_train, train_targets = x_train.iloc[train_idxs], targets.iloc[train_idxs]
    X_val, val_targets = x_train.iloc[val_idxs], targets.iloc[val_idxs]
    model, train_rmse, val_rmse = train_and_evaluate(X_train,
                                                     train_targets,
                                                     X_val,
                                                     val_targets,
                                                     max_depth=4,
                                                     n_estimators=20)
    models.append(model)
    print('Train RMSE: {}, Validation RMSE: {}'.format(train_rmse, val_rmse))
    
    
```

# Helper function to test hyperparameters with K-fold cross validation.

```python
def test_params_kfold(n_splits, **params):
    train_rmses, val_rmses, models = [],[],[]
    kfold = KFold(n_splits)
    for train_idxs, val_idxs in kfold.split(x_train):
        X_train, train_targets = x_train.iloc[train_idxs], targets.iloc[train_idxs]
        X_val, val_targets = x_train.iloc[val_idxs], targets.iloc[val_idxs]
        model, train_rmses, val_rmses = train_and_evaluate(X_train, train_targets, X_val, val_targets)
        models.append(model)
        train_rmses.append(train_rmses)
        val_rmses.append(val_rmses)
    print('Train RMSE: {}, Validation RMSE: {}'.format(np.mean(train_rmses), np.mean(val_rmses)))
    return models    
        
```
